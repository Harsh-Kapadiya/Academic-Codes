```python
import heapq

def dijkstra(graph, start, goal):
    pq = [(0, start, [])]
    seen = set()
    while pq:
        cost, node, path = heapq.heappop(pq)
        if node in seen:
            continue
        seen.add(node)
        path = path + [node]
        if node == goal:
            return path, cost
        for nxt, w in graph[node]:
            if nxt not in seen:
                heapq.heappush(pq, (cost + w, nxt, path))
    return None, float("inf")
```
