```python
# import all the necessary libraries
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns   
```

```python
# assign the dataset and drop the unnecessary columns
data = pd.read_csv('Iris.csv')
features = data.drop(columns=['Id', 'Species'], errors='ignore')
```

```python
# Initialize and fit the K-Means Model
kmeans = KMeans(n_clusters=2, random_state=42, n_init='auto')
kmeans.fit(features)
```

```python
# Save the predicted clusters back to the original DataFrame
data["Cluster"] = kmeans.labels_
```

```python
# printing the coordinates of the cluster centroids
centroids = kmeans.cluster_centers_
print("Cluster Centroids:\n", centroids)
```

```python
# output using the matlablib and seaborn libraries to visualize the clusters
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=data, 
    x="PetalLengthCm",  # Adjust column names according to your CSV header
    y="PetalWidthCm", 
    hue="Cluster", 
    palette="viridis", 
    s=70
)
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centroids')
plt.title('K-Means Clustering of Iris Dataset')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()
```
