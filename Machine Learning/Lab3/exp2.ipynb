```python
import numpy as np
import pandas as pd

path = pd.read_csv("Google_Stock_Price_Train.csv")

x = path["Open"].values
# x matrix
X_dot = np.column_stack((np.ones(len(x)), x))
Y = path["High"].values
```

```python
# transpose of x and y
X_transpose = np.transpose(X_dot)
Y_transpose = np.transpose(Y)
```

```python
theta = (Y_transpose @ X_dot)/ (X_transpose @ X_dot)
```

```python
learning_Rate = 0.00000001
itrations = 1000
```

```python
for i in range(itrations):
    Y_pred = X_dot @ theta
    gradient = (X_transpose @ Y_pred) / len(Y)
    theta = theta - learning_Rate * gradient
```

```python
print("Theta after training:", theta)
print("Predicted Volume:", Y_pred)
print("Actual Volume:", Y)
```

```python
import matplotlib.pyplot as plt
# Plot
plt.figure(figsize=(12,6))

plt.plot(Y, color='blue', label='Actual High Price')
plt.plot(Y_pred, color='red', label='Predicted High Price')

plt.title("Actual vs Predicted High Price")
plt.xlabel("Training Samples")
plt.ylabel("Stock Price")
plt.legend()

plt.grid(True)
plt.show()
```
