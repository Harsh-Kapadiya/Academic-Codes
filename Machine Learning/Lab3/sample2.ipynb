```python
import pandas as pd
import numpy as np
df = pd.read_csv("student-por.csv")
print(df.head())
```

```python
print(df.describe())
```

```python
print(df.info())
```

```python
for col in df :
    if(df[col].dtype not in ("int64", "float64")):
        df[col],_=pd.factorize(df[col])
print(df.info())
```

```python
train_data = df[: 8*len(df)//10]
test_data = df[8*len(df)//10 :]
```

```python
theta_1 = (
    np.mean(train_data["G1"] * train_data["G3"])
    - np.mean(train_data["G1"]) * np.mean(train_data["G3"])
) / (
    np.mean(np.square(train_data["G1"]))
    - np.square(np.mean(train_data["G1"]))
)

theta_0 = np.mean(train_data["G3"]) - theta_1*np.mean(train_data["G1"])
```

```python
print("theta_1 = ",theta_1)
print("theta_0 = ",theta_0)
```

```python
test_data["simple_prediction"] = theta_1 * test_data["G2"] + theta_0
test_data
```

```python
# 1. Choose your benchmarks
PASSING_MARK = 10.0

# 2. Initialize confusion matrix counters
tp, tn, fp, fn = 0, 0, 0, 0

# 3. Zip and loop through actual scores and your new "simple_prediction" values
# Replace "G3" with "G2" or "G1" if you are validating against a different target
actual_scores = test_data["G3"]
predicted_scores = test_data["simple_prediction"]

for actual, pred in zip(actual_scores, predicted_scores):
    # Convert true continuous score to binary (Pass=1, Fail=0)
    true_class = 1 if actual >= PASSING_MARK else 0
    
    # Convert your linear regression prediction to binary (Pass=1, Fail=0)
    pred_class = 1 if pred >= PASSING_MARK else 0
    
    # Sort into the four quadrants of the confusion matrix
    if true_class == 1 and pred_class == 1:
        tp += 1
    elif true_class == 0 and pred_class == 0:
        tn += 1
    elif true_class == 0 and pred_class == 1:
        fp += 1
    elif true_class == 1 and pred_class == 0:
        fn += 1

# 4. Display the results visually
print("=" * 40)
print("          CONFUSION MATRIX")
print("=" * 40)
print(f"True Negatives  (TN): {tn:4d} | False Positives (FP): {fp:4d}")
print(f"False Negatives (FN): {fn:4d} | True Positives  (TP): {tp:4d}")
print("=" * 40)

# Calculate and display Accuracy
total = tp + tn + fp + fn
accuracy = (tp + tn) / total if total > 0 else 0
print(f"Prediction Accuracy: {accuracy:.2%}\n")
```
