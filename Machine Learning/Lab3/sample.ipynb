```python
# Sample outputs from your linear regression model
predictions = [0.85, 0.23, 0.61, 0.45, 0.92, 0.12]
true_labels = [1, 0, 1, 0, 1, 1]

# 1. Define your decision boundary
threshold = 0.5

# 2. Apply threshold to get binary classes
binary_predictions = []
for p in predictions:
    if p >= threshold:
        binary_predictions.append(1)
    else:
        binary_predictions.append(0)

# Print the converted outputs
print("Binary Predictions:", binary_predictions)
```

```python
# 3. Initialize counts
tp = 0  # True Positive
tn = 0  # True Negative
fp = 0  # False Positive
fn = 0  # False Negative

# 4. Count the overlaps
for true, pred in zip(true_labels, binary_predictions):
    if true == 1 and pred == 1:
        tp += 1
    elif true == 0 and pred == 0:
        tn += 1
    elif true == 0 and pred == 1:
        fp += 1
    elif true == 1 and pred == 0:
        fn += 1

# 5. Display the 2x2 matrix
print(f"[{tn} {fp}]")
print(f"[{fn} {tp}]")
```
