```python
#Vanilla linear regression
import pandas as pd
path = pd.read_csv('Google_Stock_Price_Train.csv')
x = path.iloc[:, 1:2].values
y = path.iloc[:, 2:3].values
```

```python
# finding mean values of x and y
import numpy as np
x_mean = np.mean(x)
y_mean = np.mean(y)
```

```python
theta_1 = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
theta_0 = y_mean - (theta_1 * x_mean)
```

```python
M_0 = theta_0 + (theta_1 * x)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x, y, color='blue', label='Actual Prices')
plt.plot(x, M_0, color='red', label='Regression Line')
plt.title('Google Stock Price Prediction')
plt.xlabel('Days')
plt.ylabel('Stock Price')
plt.legend()
plt.show()
```
