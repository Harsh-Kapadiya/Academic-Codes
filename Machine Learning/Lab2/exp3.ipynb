```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
path = pd.read_csv("Google_Stock_Price_Train.csv")
x = path["Open"].values
Y = path["High"].values
```

```python
# X matrix
# First column = 1 for intercept
X = np.column_stack((np.ones(len(x)), x))
```

```python
# Transpose of X
X_transpose = np.transpose(X)
# Ridge regularization parameter
lambda_value = 1.0
# Identity matrix
I = np.eye(X.shape[1])
I[0, 0] = 0
```

```python
# Ridge Regression using Matrix Formula
theta = np.linalg.inv(
    X_transpose @ X + lambda_value * I
) @ X_transpose @ Y
print("Theta after training:")
print(theta)
# Prediction
Y_pred = X @ theta
print("Predicted High Price:")
print(Y_pred)
print("Actual High Price:")
print(Y)
```

```python
# Plot
plt.figure(figsize=(12, 6))

plt.plot(Y, label="Actual High Price")
plt.plot(Y_pred, label="Predicted High Price")
plt.title("Ridge Regression: Actual vs Predicted High Price")
plt.xlabel("Training Samples")
plt.ylabel("Stock Price")
plt.legend()
plt.grid(True)
plt.show()
```
